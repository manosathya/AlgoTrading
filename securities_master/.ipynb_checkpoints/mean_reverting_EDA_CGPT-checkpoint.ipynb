```python
%matplotlib inline
import pandas as pd
import numpy as np

import mplfinance as mpl 
import matplotlib.pyplot as plt
import seaborn as sns

from tqdm.notebook import tqdm
```

```python
from database_query import get_price_data, get_symbols_with_pd
```

```python
from joblib import Parallel, delayed
from tqdm_joblib import tqdm_joblib
import gc

import multiprocessing
from memory_profiler import profile

from statsmodels.tsa.stattools import coint, adfuller
```

```python
data = get_price_data(get_symbols_with_pd(), date_range='2024-1-1::2025-1-1', default_cols=['close'])
data.close = data.close.astype('float64')
data.set_index('price_date', inplace=True)

data = data.pivot(columns=['ticker']).dropna()
data
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(data.corr(), annot=True)
plt.show()
```

```python
#%%time
tickers = data.close.columns.values
pairs = [(i,j) for k,i in enumerate(tickers) for j in tickers[k+1:] if i!=j]

def coint_df(i,j):
    ts1 = data.close.loc[:,i].values
    ts2 = data.close.loc[:,j].values
    res = coint(ts1,ts2)
    del ts1, ts2
    gc.collect()
    return res
    
#with tqdm_joblib(total=len(pairs)) as progress_bar:
results = Parallel(n_jobs=3, verbose=55)(delayed(coint_df)(i,j) for i,j in pairs)

#cointegration = pd.DataFrame(columns=tickers,  index=tickers)
#for i,j in tqdm(pairs):
    #cointegration.loc[i, j] = coint_df(i,j)[1]
#cointegration
```

```python
@profile
def adf(i):
    ts1 = data.close.loc[:,i].values
    res = adfuller(ts1)
    del ts1
    gc.collect()
    return res

#results = Parallel(n_jobs=6, batch_size=1, verbose=50)(delayed(adf)(i) for i in tickers)
adf('AAPL')
```
